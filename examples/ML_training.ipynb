# iDILI Model Training Example

Long-to-wide data is used to select a binary classifier using [AutoGluon](https://auto.gluon.ai/stable/index.html). Quality metrics are recorded for the leader of the models and then the whole dataset is scored and the probability for the predictions are appened to the long-to-wide data. Feature imbalance is handled by [imblearn SMOTE](https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTE.html) on the training set

## Quality Metrics

- Accurarcy
- f1-pos/f1-neg
- Final model loss
- ROC AUC
- Confusion Matrix

```python
import os
import re

import pickle

import numpy as np
import pandas as pd

from idili.train_models import train_model

with open('example_data/wide_data.pkl', 'rb') as f:
    data_cols = pickle.load(f)
with open('example_data/wide_meta.pkl', 'rb') as f:
    meta_cols = pickle.load(f)

df = pd.read_csv('example_data/df_wide.csv')
```

```python
model_dir = 'example_model/'
cell_line='723'

train_model(df, data_cols=data_cols, meta_cols=meta_cols, model_dir=model_dir, cell_line=cell_line)
```
